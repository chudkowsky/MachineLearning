```python
import numpy as np
import pandas as pd
import sklearn as sl
import matplotlib as mpl
import matplotlib.pyplot as plt
import seaborn as sns
sns.set()
from sklearn import svm
from sklearn.datasets import load_diabetes
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_absolute_error, mean_squared_error, root_mean_squared_error, r2_score
from sklearn.model_selection import cross_val_predict, cross_val_score,cross_validate
```

```python
data = load_diabetes(as_frame=True)
X = data.data
y = data.target

lr = LinearRegression()
sc = cross_val_score(lr, X, y, )
print(sc)
print(sc.mean())
print(sc.std())
```

```python
# Evaluate the SVR model using cross-validation on the entire dataset
cv_scores = cross_validate(lr, X, y, scoring=('r2', 'neg_mean_absolute_error', 'neg_mean_squared_error', 'neg_root_mean_squared_error'),return_train_score=True )

# Print the mean scores from cross-validation
print(f"Mean R-squared (Cross-validation): {cv_scores['test_r2'].mean()}")
print(f"Mean R-squared (Cross-validation): {cv_scores['train_r2'].mean()}")


print(f"Mean Absolute Error (Cross-validation): {-cv_scores['test_neg_mean_absolute_error'].mean()}")
print(f"Mean Absolute Error (Cross-validation): {-cv_scores['train_neg_mean_absolute_error'].mean()}")

print(f"Mean Squared Error (Cross-validation): {-cv_scores['test_neg_mean_squared_error'].mean()}")
print(f"Mean Squared Error (Cross-validation): {-cv_scores['train_neg_mean_squared_error'].mean()}")

print(f"Mean Root Mean Squared Error (Cross-validation): {-cv_scores['test_neg_root_mean_squared_error'].mean()}")
print(f"Mean Root Mean Squared Error (Cross-validation): {-cv_scores['train_neg_root_mean_squared_error'].mean()}")
```

```python
from sklearn.preprocessing import PolynomialFeatures

poly2 = PolynomialFeatures(degree=2)
X_poly2 = poly2.fit_transform(X)
X2 = pd.DataFrame(X_poly2, columns=poly2.get_feature_names_out(X.columns))

poly3 = PolynomialFeatures(degree=3)
X_poly3 = poly3.fit_transform(X)
X3 = pd.DataFrame(X_poly3, columns=poly3.get_feature_names_out(X.columns))
print("Shape of X_poly2:", X_poly2.shape)
print("Shape of X_poly3:", X_poly3.shape)
```

```python
lr2 = LinearRegression()
lr3 = LinearRegression()
cv_scores_poly2 = cross_validate(lr2, X2, y, scoring=('r2', 'neg_mean_absolute_error', 'neg_mean_squared_error', 'neg_root_mean_squared_error'), return_train_score=True)
cv_scores_poly3 = cross_validate(lr3, X3, y, scoring=('r2', 'neg_mean_absolute_error', 'neg_mean_squared_error', 'neg_root_mean_squared_error'), return_train_score=True)

print("Cross-validation scores for PolynomialFeatures degree 2:")
print(f"Mean R-squared test: {cv_scores_poly2['test_r2'].mean()}")
print(f"Mean R-squared train: {cv_scores_poly2['train_r2'].mean()}")


print(f"Mean Absolute Error: {-cv_scores_poly2['test_neg_mean_absolute_error'].mean()}")
print(f"Mean Squared Error: {-cv_scores_poly2['test_neg_mean_squared_error'].mean()}")
print(f"Mean Root Mean Squared Error: {-cv_scores_poly2['test_neg_root_mean_squared_error'].mean()}")

print("\nCross-validation scores for PolynomialFeatures degree 3:")

print(f"Mean R-squared: {cv_scores_poly3['test_r2'].mean()}")
print(f"Mean R-squared: {cv_scores_poly3['train_r2'].mean()}")

print(f"Mean Absolute Error: {-cv_scores_poly3['test_neg_mean_absolute_error'].mean()}")
print(f"Mean Squared Error: {-cv_scores_poly3['test_neg_mean_squared_error'].mean()}")
print(f"Mean Root Mean Squared Error: {-cv_scores_poly3['test_neg_root_mean_squared_error'].mean()}")
```
