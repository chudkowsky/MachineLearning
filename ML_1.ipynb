```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
sns.set()
```

```python
df = pd.read_csv('reglin_1.csv')
df.info()
df.head()
```

```python
sns.scatterplot(df, x='x', y='y');
```

```python
Sx = df['x'].sum()
Sy = df['y'].sum()
Sx2 = (df['x']**2).sum()
Sxy = (df['x']*df['y']).sum()
m = len(df)
```

```python
W = m*Sx2 - Sx**2
Wa = m*Sxy - Sx*Sy
Wb = Sx2*Sy - Sx*Sxy
```

```python
a = Wa/W
b = Wb/W
a, b
```

```python
MAE = np.abs(a*df['x']+b - df['y']).mean()
MSE = np.square(a*df['x']+b - df['y']).mean()
RMSE = MSE**(1/2)

ym = df['y'].mean()
SS1 = np.square(a*df['x']+b - ym).sum()
SS2 = np.square(df['y']- ym).sum()
R2 = SS1/SS2

print("mae: ",MAE)
print("mse: ", MSE)
print("rmse: ", RMSE)
print("r2: ", R2)
```
